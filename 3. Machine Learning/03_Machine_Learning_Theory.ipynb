## Data, signal, and noise

If a model is powerful enough, it will be able to memorize the data it was trained on and give perfect predictions for the training set.

But what's wrong with memorization (given that the goal is to build a pattern recognition algorithm)? Let's look at an example.

👩‍🏫 In this image, which model would you consider to be accurate, or the most useful? 

<img src='images/03_Machine_Learning_Theory/data-fit.png' width=800px style="margin-left:auto; margin-right:auto"> 

[source](https://inria.github.io/scikit-learn-mooc/overfit/overfitting_vs_under_fitting_slides.html)

In machine learning, we often work with data that consists of a signal and noise. The signal represents the underlying patterns or information we want to capture or model, while the noise represents random variations or irrelevant factors.

💡 Have you ever tried to **tune** an analog radio to detect a radio signal, developing or building a ML model is similar to that. You need to find the right frequency to capture the signal, while ignoring the noise. 

The art of building a good ML model is being able to capture enough of the signal, while ignoring irrelevant noise. That is, the model should be able to generalize well to new, unseen data. 

### Underfitting

If a model exists that fails to capture either signal or noise, the model is said to **underfit**.

<img src='images/03_Machine_Learning_Theory/underfit.png' width=600px style="margin-left:auto; margin-right:auto"> 

[source](https://inria.github.io/scikit-learn-mooc/overfit/overfitting_vs_under_fitting_slides.html)

In this is the case, your model is said to contain **high bias** (see notes below for more). 

### Overfitting

The opposite of underfitting is overfitting. Overfitting is when the model does capture the underlying signal, but also some of the noise. 

<img src='images/03_Machine_Learning_Theory/overfit.png' width=600px style="margin-left:auto; margin-right:auto"> 

[source](https://inria.github.io/scikit-learn-mooc/overfit/overfitting_vs_under_fitting_slides.html)

In this context, your model is said to contain **high variance** (see notes below for more).

#### Key takeaways

Models too complex for the data overfit:
- they memorize the training set
- they do not generalize to the test set

Models too simple for the data underfit:
- they fail to capture any underlying signal (and/or noise)
- they do not generalize to the test set

How to find the right trade-off, ensuring that the model generalizes well to new data?

> A model is said to be complex (for the task at hand) if it has more flexibility than needed to capture the underlying signal in the data. Depending on the model class, complexity can be controlled by the number of parameters, the degree of the polynomial, the depth of the tree, the number of layers in a neural network etc.

<img src="images/03_Machine_Learning_Theory/under-over.png" width=600px style="margin-left:auto; margin-right:auto">

🔔 **Underfitting** is when the model performs poorly on both the training set and the test set.

🔔 When a model is **overfitting**, it performs well on the *training* set, but not on the *testing* set.

# Summary

- The goal of a machine learning model is to generalize well to new, unseen data.

- Complexity, or flexibility, of a model is controlled by the number of parameters, the degree of the polynomial, the depth of the tree, the number of layers in a neural network etc.

- If a model performs badly on the train set, you know it is **underfitting**.

- If a model performs well on the train set, but badly on the test set it is **overfitting**.

🔔 The essence of machine learning is that it is an iterative process. You can start with a simple model, evaluate its performance, and then try more complex models. Remember to put on your experimenter hat and try different models to see what works best for your data.

### Evaluating the generalization of ML models

To evaluate how well can our model generalize to new data, one common technique is to split the available data into training and testing sets. This technique is known as train-test split. 

<img src="images/03_Machine_Learning_Theory/train-test.png" width="400" style="margin-left:auto; margin-right:auto">

The purpose of train-test split is to evaluate the performance of the model on unseen data. By training the model on the training set and evaluating it on the testing set, we can assess how well the model generalizes to new data. 

🔔 If the model performs poorly on both the training and testing sets, it's possibly underfitting.

🔔 If the model performs well on the training set but poorly on the testing set, it's possibly overfitting.

Another method to evaluate the generalization of the model is to use **cross-validation**. You will learn more about cross-validation next week.


<details>
  <summary>The bias-variance tradeoff </summary>
  <img src='images/03_Machine_Learning_Theory/bias-variance.svg' width=800px align='left'> <br>
  
  The bias-variance tradeoff represent the theoretical underpinnings of underfitting and overfitting, and explains why a lot of ML models tend to do well in practice. It describes the tradeoff between the bias of the model and its variance. Bias refers to the error introduced by approximating a real-world problem, which requires a complex model, by a simple model. Variance refers to the error introduced by the model's sensitivity to fluctuations (noise) in the training set (often due to using a model more complex than required). The goal is to find a model that minimizes both bias and variance. Hence, a model that can generalize well to unseen data. while capturing the underlying signal in the data.

### Regression models

#### Linear Regression

The simplest model for regression is linear regression. In linear regression, the model assumes that the relationship between the input features and the target variable is linear.

For example, trying to predict the penguins body mass ($y$), from their bill depth ($x$):

$$ y = β_0 + β_1x $$

Here $β_0$ is the intercept and $β_1$ is the slope of the line

🔔 The intercept $β_0$ is the value of $y$ when $x=0$ i.e. the point where the line crosses the y-axis. 

🔔 The slope $β_1$ is the change in $y$ for a unit change in $x$.

````python
import seaborn as sns
import pandas as pd

penguins = pd.read_csv('data/penguins.csv')

def simple_regression():
    sns.regplot(x=penguins['bill_length_mm'], y=penguins['body_mass_g'],
            scatter_kws={'s': 10, 'alpha': 0.2}, 
            line_kws={"color": "red", 'linewidth':2},
            ci=None)
    
simple_regression()
````

Regression can also be extended to multiple features. For example, trying to predict the penguins body mass ($y$), from their bill depth ($x_1$), flipper length ($x_2$) and bill_length ($x_3$):

$$ y = β_0 + β_1x_1 + β_2x_2 + β_3x_3 $$

Or regression can be extended to polynomial features. For example, trying to predict the penguins body mass ($y$), from their bill depth ($x$) using a polynomial of degree 3:

$$ y = β_0 + β_1x + β_2x^2 + β_3x^3 $$

````python
def poly_regression():
    sns.regplot(x=penguins['bill_length_mm'], y=penguins['body_mass_g'],
            scatter_kws={'s': 10, 'alpha': 0.2}, 
            line_kws={"color": "red", 'linewidth':2},
            order=3,
            ci=None)
    
poly_regression()
````

### Ensemble methods

A shallow decision tree is a simple model that is likely to underfit the data. A deep decision tree is a complex model that is likely to overfit the data.
Ensemble models try to find a balance between underfitting and overfitting by combining multiple (deep) decision trees. 

🔔 The reason behind the unbearable success of ensemble methods is that they average out the predictions of multiple models. When we average out the variances of multiple models, the resulting variance is lower than the variance of an individual model (hurray mathematics!). 

The two most popular ensemble methods are bagging and boosting:
- **Bagging**: Training a bunch of individual models in a *parallel* way. Each model is trained by a random subset of the data
- **Boosting**: Training a bunch of individual models in a *sequential* way. Each individual model learns from mistakes made by the previous model.

#### Random Forest (bagging)

A Random Forest ensembles a large number of independent unrestricted decision trees. The trees are trained on (bootstrapped) subsamples of the training data, with a random selection of features.

After each tree produces a prediction, voting or averaging is used to produce the final prediction.

### k-Nearest Neighbors (kNN)

<img src="images/03_Machine_Learning_Theory/KnnClassification.png" width="600" style="margin-left:auto; margin-right:auto">

🔔 The value of k controls the flexibility of the model. A small value of k will lead to a complex model and likely overfit, while a large value of k will lead to a simple model, and likely underfit.

Data points that are close to each other are more similar than data points that are far apart. Distance-based models use this principle to make predictions. One of the most popular distance-based models is the k-Nearest Neighbors (kNN) algorithm.

The kNN algorithm classifies a new data point based on the majority class of its k-nearest neighbors. The distance between data points is calculated using a distance metric, such as Euclidean distance.

### Support Vector Machine

SVMs use hyperplanes to separate binary datasets into two classes. 🔔 A hyperplane is a surface that divides a space into two parts.

Examples:
- In 2D (like a piece of paper), a hyperplane is a straight line.
- In 3D (like a room), a hyperplane is a flat plane.
- In higher dimensions, it’s a similar flat surface that extends through that space.
