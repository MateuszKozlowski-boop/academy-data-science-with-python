# Need for interpretability

**Interpretability** is the degree to which a model can be understood in human terms.

<mark>**Question**: Can you think of 2 situations where interpretability of a model is important?</mark>

- Why was this customer denied a loan? 
- Customer churn: can we identify why customers leave, and therefore stop them leaving in the future?

<mark>**Question:** What are some models that are inherently more interpretable? How about models that are very hard to interpret?</mark>

<img src='images/05_Choosing_Models/black-box.svg' width=500px align=right>

### Examples: 
**Easy to interpret:**
- Linear and logistic regression
- Decision Trees

**Hard to interpret:**
- Support Vector Machines
- Neural Networks

<mark>**Question:** What are exampes where only high performance is important?</mark>

### Examples:
- Handwriting recognition
- Predicting stock prices

When you only care about high performance, it's usually best to choose a more flexible model. These models are often less interpretable.

# Implementation Costs

Implementation cost can for example depend on:
- The training time of the algorithm (especially when large training datasets are required)
- Required domain expertise: you may need more knowledge to make some models work

**Easier to implement:**
- Linear regression
- Logistic regression
- Decision Tree

**Harder to implement:**
- Neural networks
- Support vector machine
- Random forests

# Number of Features

The dataset may have a large number of features that may not all be relevant. For genetic or textual data, the number of features can also be very large compared to the number of data points.

Algorithms that are negatively impacted by a large number of features are for example K-Nearest Neighbour and Logistic Regression. Suppory Vector Machines are a better choice in that case. 

<mark>**Question:** What are other things you could do besides choosing a different algorithm when you have a large number of features?</mark>

If you have too many features, principal component analysis (PCA) and feature selection techniques can be used to reduce dimensionality and select important features without losing information.
